```python
bank_churned = pd.merge(
    customer_info,
    account_info,
    on= "CustomerId",
    how="inner"

)
```

```python
bank_churned =  bank_churned.dropna()
print(f"Null values successfully dropped")
```

```python
bank_churned.info()
```

```python
bank_churned = bank_churned.drop(columns= ['Tenure_x'])
```

```python
bank_churned.info()
```

```python
bank_churned.head(5)
```

```python
bank_churned.isnull().sum()
```

```python
#female = 0, amle = 1,
bank_churned['Gender'] = bank_churned['Gender'].map({
    'Female' : 0,
    'Male' : 1
})
print(bank_churned['Gender'].dtype)
```

```python
bank_churned['IsActiveMember'] = bank_churned['IsActiveMember'].map({
  'Yes': 1,
  'No': 0
})
print(bank_churned['IsActiveMember'].dtype)
```

```python
bank_churned['HasCrCard'] = bank_churned['HasCrCard'].map({
    'Yes': 1,
    'No' : 0
})
print(bank_churned['HasCrCard'].dtype)
```

```python
#Remove the currency symbol and convert to float
bank_churned['EstimatedSalary'] = bank_churned['EstimatedSalary'].str.replace("€", "", regex=False).astype(float)
print(bank_churned['EstimatedSalary'].dtype)
```

```python
#change datatype
bank_churned['Balance'] = bank_churned['Balance'].str.replace("€", "", regex=False).astype(float)
print(bank_churned['Balance'].dtype)
```

ML MODEL USING DECISION TREE


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
#predictive feature
y = bank_churned.Exited
print(y.head(5))
```

```python
#SELECTING FEATURES
churning_features  = ['Gender', 'Age', 'EstimatedSalary', 'Balance', 'NumOfProducts', 'HasCrCard', 'Tenure_y', 'IsActiveMember']
X = bank_churned[churning_features]
print(X.head(5))
print(X.dtypes)
```

```python
#train_test_split
train_X, val_x, train_y,val_y =train_test_split(X, y, random_state=42)
```

```python
#Dpecify and define model
churning_model = DecisionTreeClassifier(random_state=42, class_weight= 'balanced')
churning_model.fit(train_X,train_y)
```

```python
#predict with all validation observations
val_predictions = churning_model.predict(val_x)
```

```python
#print the top val_predictions
print(churning_model.predict(val_x.head(5)))
#print actual states of members from validation data
print(bank_churned['Exited'].head(5))
```

Measures the percentage of predictions your model got correct.

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

accuracy = accuracy_score(val_y, val_predictions)
confusion = confusion_matrix(val_y, val_predictions)

print(f'Accuracy: {accuracy}')
print(f'confusion: {confusion}')
```

```python
from sklearn.metrics import classification_report
print(classification_report(val_y, val_predictions))
```
